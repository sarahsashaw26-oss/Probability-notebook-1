```python
import pandas as pd
```

```python
titanic = pd.read_csv('titanic.csv')
titanic.head()

data = titanic[['survived', 'pclass', 'sex', 'age', 'alive', 'alone', 'embark_town', 'class']].copy()
data
```

```python
data.shape[0]
```

```python
alone = data["alone"] == True
survived = data["survived"] == 1
```

```python
def f(condition):
    return condition.mean()
```

```python
f(alone)
```

```python
f(survived)
```

## Joint Events
These are events that occur at the same time.\
Example:\
How many passengers embarked on southampton and travelled alone


```python
embark_southampton = data['embark_town'] == 'Southampton'
alone = data['alone'] == True

joint_event = embark_southampton & alone
joint_event.sum()
```

what is the probability of a passenger chosen at random who embarked at southampton and travelled alone

```python
prob = f(embark_southampton & alone)
print (f"The probability is {prob:.4f}")
```

#### The Contingency Table(Frequency Table)

```python
joint = pd.crosstab(data['sex'], data['survived'])
joint.columns = ['did not survive', 'survived']
joint
```

```python
joint = pd.crosstab(data['sex'], data['survived'], normalize=True)
joint.columns = ['did not survive', 'survived']
joint
```
